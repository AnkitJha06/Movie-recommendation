```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
movies_data = pd.read_csv('movies.csv')
ratings_data = pd.read_csv('ratings.csv')
links_data = pd.read_csv('links.csv')
tags_data = pd.read_csv('tags.csv')
```

```python
for name, df in {
    "Movies": movies_data,
    "Ratings": ratings_data
}.items():

    print(f"\n{'=' * 15} {name} {'=' * 15}")

    print("\nShape:", df.shape)

    print("\nColumn types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nExact duplicate rows:")
    print(df.duplicated().sum())

    print("\nSample:")
    print(df.head())
```

```python
print("Unique movie IDs in movies:",
      movies_data["movieId"].nunique())

print("Duplicate movie IDs in movies:",
      movies_data["movieId"].duplicated().sum())

print("Unique users:",
      ratings_data["userId"].nunique())

print("Unique rated movies:",
      ratings_data["movieId"].nunique())

print("\nRating distribution:")
print(ratings_data["rating"].value_counts(dropna=False).sort_index())

print("\nRating summary:")
print(ratings_data["rating"].describe())

# Expected rating scale for this project
valid_ratings = [0.5, 1.0, 1.5, 2.0, 2.5,
                 3.0, 3.5, 4.0, 4.5, 5.0]

invalid_ratings = ratings_data[
    ~ratings_data["rating"].isin(valid_ratings)
]

print("\nInvalid or missing ratings:", len(invalid_ratings))
print(invalid_ratings.head(10))

# Ratings referring to movies absent from the catalogue
unknown_movie_ratings = ratings_data[
    ~ratings_data["movieId"].isin(movies_data["movieId"])
]

print("\nRatings with unknown movie IDs:",
      len(unknown_movie_ratings))

print(unknown_movie_ratings.head(10))

# Repeated user–movie pairs, regardless of rating/timestamp
repeated_pairs = ratings_data[
    ratings_data.duplicated(
        subset=["userId", "movieId"],
        keep=False
    )
].sort_values(["userId", "movieId"])

print("\nRows belonging to repeated user–movie pairs:",
      len(repeated_pairs))

print(repeated_pairs.head(10))

unrated_movies = movies_data[
    ~movies_data["movieId"].isin(ratings_data["movieId"])
]

print("\nMovies without ratings:", len(unrated_movies))
```

```python
ratings_per_user = ratings_data.groupby("userId").size()
ratings_per_movie = ratings_data.groupby("movieId").size()

print("Ratings per user:")
print(ratings_per_user.describe())

print("\nRatings per movie:")
print(ratings_per_movie.describe())

print("\nMovies with fewer than 5 ratings:",
      (ratings_per_movie < 5).sum())

print("\nUsers with fewer than 5 ratings:",
      (ratings_per_user < 5).sum())

numeric_timestamps = pd.to_numeric(
    ratings_data["timestamp"],
    errors="coerce"
)

rating_dates = pd.to_datetime(
    numeric_timestamps,
    unit="s",
    utc=True,
    errors="coerce"
)

print("\nInvalid or missing timestamps:",
      rating_dates.isna().sum())

print("\nEarliest rating:", rating_dates.min())
print("Latest rating:", rating_dates.max())
```
